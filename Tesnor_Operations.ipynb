```python
import torch
```

```python
my_tensor = torch.arange(10)
my_tensor
```

```python
# Reshape and View

my_tensor_r = my_tensor.reshape(2,5)

print("Reshape: ", my_tensor_r)

my_tensor_v = my_tensor.view(2,5)
print("View: ", my_tensor_v)
```

```python
# Reshape using -1

my_tensor1 = torch.arange(10)
my_tensor1 = my_tensor.reshape(-1,5)
my_tensor1
```

```python
# Reshape and View update the tensor

# An advantage of using tensors is: when you update the original tensor
# the reshaped tensor will be updated

tensor2 = torch.arange(10)
print("Tesnor 2: ", tensor2)

tensor3 = tensor2.reshape(5, 2)
print("Tensor 3: ", tensor3)

tensor2[0] = 100
print("Tensor 2 after update: ", tensor2)
print("Tensor 3 after update: ", tensor3)
```

```python
# Slices

# A specific item
tensor2[1] = 5
print(tensor2[1])


# Grap a slice with keeping the shape
tensor3[:, 1:]
```
